```python
import pandas as pd
import matplotlib.pyplot as plt
print("Libraries installed successfully!")

url = 'https://raw.githubusercontent.com/sophiaAcademics/BDA_Excel/main/StudentData/Assessments/Unit6/TechGearTouchstone/techgear_sales_data.xlsx'
sales_tech = pd.read_excel(url)

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
import numpy as np

features = ['Ad_Spend_Facebook', 'Ad_Spend_Instagram', 'Discount_Rate']
target = 'Sales'

x = sales_tech[features]
y = sales_tech[target]

model = LinearRegression()
tree_model = DecisionTreeRegressor()

np.random.seed(42)
mse_scores = cross_val_score(model, x, y, cv=5, scoring='neg_mean_squared_error')
rmse_scores = (-mse_scores) ** 0.5
average_rmse = rmse_scores.mean()
print(f"Average Root Mean Squared Error from 5-fold Cross-Validation for a Multiple Linear Regression model: {average_rmse:.0f}")

mse_scores = cross_val_score(tree_model, x, y, cv=5, scoring='neg_mean_squared_error')
rmse_scores = (-mse_scores) ** 0.5
average_rmse = rmse_scores.mean()
print(f"Average Root Mean Squared Error from 5-fold Cross-Validation for a Decision Tree Model: {average_rmse:.0f}")

model.fit(x, y)

print(model.coef_)
print(model.intercept_)
```

```python
url = 'https://raw.githubusercontent.com/sophiaAcademics/BDA_Excel/main/StudentData/Assessments/Unit6/TechGearTouchstone/techgear_sales_data_monthly.xlsx'
sales_tech = pd.read_excel(url)

import pandas as pd
from statsmodels.tsa.holtwinters import SimpleExpSmoothing

# 3-month Moving Average
sales_tech['TECH_3'] = sales_tech['Sales'].rolling(window=3).mean()
print(sales_tech['TECH_3'].tail(5))

ma_data = sales_tech.dropna(subset=['TECH_3'])
print("=== Moving Average (TECH_3) predictions ===")
print(sales_tech['TECH_3'].dropna().tail(3))

import numpy as np

# Moving Average errors
ma_data = sales_tech.dropna(subset=['TECH_3'])
ma_errors = ma_data['Sales'] - ma_data['TECH_3']
mse_ma = np.mean(ma_errors ** 2)
mae_ma = np.mean(np.abs(ma_errors))
mape_ma = np.mean(np.abs(ma_errors / ma_data['Sales'])) * 100

print("=== 3-Month Moving Average ===")
print(f"MSE:  {mse_ma:.1f}")
print(f"MAE:  {mae_ma:.1f}")
print(f"MAPE: {mape_ma:.1f}%")

# Simple Exponential Smoothing
ses_model = SimpleExpSmoothing(sales_tech['Sales']).fit(
    smoothing_level=0.8,
    optimized=False
)

print("=== Simple Exponential Smoothing predictions ===")
sales_tech['SES'] = ses_model.fittedvalues
print(sales_tech['SES'].tail(5))

ses_errors = sales_tech['Sales'] - sales_tech['SES']

mse_ses = np.mean(ses_errors ** 2)
mae_ses = np.mean(np.abs(ses_errors))
mape_ses = np.mean(np.abs(ses_errors / sales_tech['Sales'])) * 100

print("=== Simple Exponential Smoothing ===")
print(f"MSE:  {mse_ses:.1f}")
print(f"MAE:  {mae_ses:.1f}")
print(f"MAPE: {mape_ses:.1f}%")

print(sales_tech[['Sales', 'TECH_3', 'SES']].tail(3))
```
